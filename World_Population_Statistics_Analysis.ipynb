```python

import numpy as np
import pandas as pd

# CSV file ko read kar ke DataFrame mein store kar rahe hain
pd = pd.read_csv('world_population_statistics_intermediate.csv')

# CSV se har column ka data alag variables mein nikal rahe hain
countries = pd["Country"]
year = pd["Year"]
population = pd["Population_Millions"]
growth = pd["Annual_Growth_Rate_Pct"]
birth_rate = pd["Birth_Rate_per_1000"]
death_rate = pd["Death_Rate_per_1000"]
urban = pd["Urban_Population_Pct"]
density = pd["Population_Density_per_km2"]
median_age = pd["Median_Age_Years"]
fertility = pd["Fertility_Rate"]


# 1. General Statistics
# Population data ki basic statistics calculate kar rahe hain
def general_statistics():
    print("="*40)
    print("====GENERAL STATISTICS=====")

    # Total countries ki tadaad count kar rahe hain
    print("Total Countries:", len(countries))

    # Total, average aur median population calculate kar rahe hain
    print("Total Population:", round(np.sum(population),2),"millions")
    print("Average Population:", round(np.mean(population),2),"millions")
    print("Median Population:", round(np.median(population),2), "millions")

    # Sab se zyada aur sab se kam population nikal rahe hain
    print("Highest Population:", np.max(population),"millions")
    print("lowest Population:", np.min(population), "millions")

    # Standard deviation se population values ka phailao dekh rahe hain
    print("Population Dev Std:", round(np.std(population),2))


# 2. Top 10 Populated Countries
# Sab se zyada population wale top 10 countries dhoond rahe hain
def top_populated_countries():
    print("====TOP 10 POPULATED COUNTRIES====")

    # Population ko bara se chhota sort kar ke top 10 indexes le rahe hain
    order = np.argsort(population)[::-1][:10]

    # Har country ka rank, naam aur population print kar rahe hain
    for rank, i in enumerate(order, start=1):
        print(
            rank, countries[i],
            "-", population[i], "million"
        )


# 3. Population Growth Analysis
# Countries ki population growth ko categories mein divide kar rahe hain
def growth_analysis():
    print("=====POPULATION GROWTH ANALYSIS=====")

    # 2% se zyada, zero se kam aur 0 se 2% tak growth wali countries dhoond rahe hain
    fast_growth = growth > 2
    negative_growth = growth < 0
    moderate_growth = (growth >= 0) & (growth <= 2)

    # Har category mein countries ki tadaad count kar rahe hain
    print("Fastest Growth:", np.sum(fast_growth))
    print("Moderate Gorwth:", np.sum(moderate_growth))
    print("Negative Growth:", np.sum(negative_growth))

    # 2% se zyada growth wali countries print kar rahe hain
    print("Countries with fastest growth above 2%:")
    print(countries[fast_growth])

    # Negative growth wali countries print kar rahe hain
    print("Countries with negative growth:")
    print(countries[negative_growth])


# 4. Urban Population Analysis
# Shehri ilaqon mein rehne wali abadi ka analysis kar rahe hain
def urban_analysis():
    print("="*40)
    print("====URBAN POPULATION ANALYSIS=====")

    # Average urban population percentage calculate kar rahe hain
    print("Average Urban Population:")
    print(round(np.mean(urban),2))

    # 70% se zyada urban population wali countries select kar rahe hain
    highly_urban = urban > 70

    print("Urban countries population with above 70%:")
    print(countries[highly_urban])


# 5. Birth and Death Rate Analysis
# Birth aur death rates ka comparison kar rahe hain
def birth_death_analysis():
    print("="*40)
    print("====BIRTH AND DEATH RATE ANALYSIS=====")

    # Average birth aur death rates calculate kar rahe hain
    print("Average Birth Rate:", round(np.mean(birth_rate),2))
    print("Average Death Rate:", round(np.mean(death_rate),2))

    # Birth rate mein se death rate minus kar ke natural rate nikal rahe hain
    natural_rate = birth_rate - death_rate

    print("Average Natural Rate:")
    print(round(np.mean(natural_rate),2), "per 1,000 people")

    # 30 se zyada birth rate wali countries select kar rahe hain
    hihgly_birth = birth_rate > 30
    print("Countries with above rate 30%:")
    print(countries[hihgly_birth])


# 6. Population Density Analysis
# Har area mein population ke phailao ka analysis kar rahe hain
def density_analysis():

    print("="*40)
    print("=====DENSITY ANALYSIS=====")

    # Tamam countries ki average population density calculate kar rahe hain
    print("Average Density:", round(np.mean(density),2))

    # Density ko descending order mein sort kar ke top 5 indexes le rahe hain
    order = np.argsort(density)[::-1][:5]

    print("Top 5 countries by density:")

    # Top 5 countries ki density print kar rahe hain
    for i in order:
        print(countries[i], "-", density[i], "people/km2")


# 7. Age and Fertility Analysis
# Median age aur fertility rate ka analysis kar rahe hain
def age_fertility_analysis():

    print("="*40)
    print("=====AGE AND FERTILITY====")

    # Average median age aur fertility rate calculate kar rahe hain
    print("Average Median Age:", round(np.mean(median_age),2))
    print("Average fertility Rate:", round(np.mean(fertility),2))

    # 25 saal se kam median age wali countries select kar rahe hain
    young_age = median_age < 25
    print("Countries with below age 25:")
    print(countries[young_age])

    # 4 se zyada fertility rate wali countries select kar rahe hain
    hihgly_fertility = fertility > 4
    print("Countries with above age 4:")
    print(countries[hihgly_fertility])


# 8. Population Forecast
# Growth rate ki madad se aglay saal ki estimated population nikal rahe hain
def population_forecast():

    print("="*40)
    print("====NEXT YEAR ESTIMATE====")

    # Current population par annual growth rate apply kar rahe hain
    estimate_population = population * (1 + growth / 100)

    # Estimated population ko sort kar ke top 10 indexes le rahe hain
    order = np.argsort(estimate_population)[::-1][:10]

    # Har selected country ki current aur estimated population print kar rahe hain
    for i in order:
        print(
            countries[i],
            "|Curennt:", "-", population[i], "millions",
            "|Estimate:", round(estimate_population[i],2),
        )


# 9. Correlation Analysis
# Do variables ke darmiyan relationship check kar rahe hain
def correlation_analysis():
    print("="*40)
    print("=====COOLERATION ANALYSIS=====")

    # Population aur growth rate ke darmiyan correlation nikal rahe hain
    pop_growth_corr = np.corrcoef(population, growth)[0,1]

    # Birth rate aur fertility rate ka correlation nikal rahe hain
    birth_fertility_corr = np.corrcoef(birth_rate, fertility)[0,1]

    # Urban population aur density ka correlation nikal rahe hain
    urban_density_corr = np.corrcoef(urban, density)[0,1]

    # Correlation values ko 3 decimal places tak print kar rahe hain
    print("Population Vs Growth:", round(pop_growth_corr,3))
    print("Birth Rate Vs Fertility:", round(birth_fertility_corr,3))
    print("Urban Vs Density:", round(urban_density_corr,3))


# 10. Search Country
# User ke diye gaye naam se country ka data search kar rahe hain
def country_search():

    print("=====SEARCH COUNYTRY=====")

    # User se country ka naam le rahe hain
    # strip() shuru aur aakhir ki extra spaces hatata hai
    # casefold() letters ke uppercase/lowercase ka farq khatam karta hai
    name = input("ENTER COUNTRY NAME:").strip().casefold()

    # Country names ko lowercase mein convert kar ke user ke input se match kar rahe hain
    matches = np.char.lower(countries.to_numpy().astype(str)) == name

    # Check kar rahe hain ke koi country match hui hai ya nahi
    if np.any(matches):

        # Match hone wali country ka index nikal rahe hain
        i = np.where(matches)[0][0]

        # Selected country ki statistics print kar rahe hain
        print("Country:", countries[i])
        print("Population:", population[i],"millions")
        print("Growth Rate:", growth[i], "%")
        print("Urban Population:", urban[i],"%")
        print("Population Density:", density[i],)
        print("Median Age:", median_age[i],)
        print("Fertility Rate:", fertility[i],)

    # Agar country match na ho to ye message print hoga
    else:
        print("Country Not Found!")


# Main Menu
# Menu ko baar baar chala rahe hain jab tak user exit select na kare
while True:
    print("\n===== WORLD POPULATION ANALYZER =====")
    print("1. General Statistics")
    print("2. Top 10 Populated Countries")
    print("3. Growth Analysis")
    print("4. Urban Population Analysis")
    print("5. Birth and Death Analysis")
    print("6. Population Density")
    print("7. Age and Fertility Analysis")
    print("8. Population Forecast")
    print("9. Correlation Analysis")
    print("10. Search Country")
    print("0. Exit")

    # User se menu option le rahe hain aur extra spaces hata rahe hain
    choice = input("Select an option: ").strip()

    # User ke select kiye gaye option ke mutabiq function chala rahe hain
    if choice == "1":
        general_statistics()
    elif choice == "2":
        top_populated_countries()
    elif choice == "3":
        growth_analysis()
    elif choice == "4":
        urban_analysis()
    elif choice == "5":
        birth_death_analysis()
    elif choice == "6":
        density_analysis()
    elif choice == "7":
        age_fertility_analysis()
    elif choice == "8":
        population_forecast()
    elif choice == "9":
        correlation_analysis()
    elif choice == "10":
        country_search()
    elif choice == "0":
        # Program ko khatam karne ka message print kar rahe hain
        print("Bhai, analysis complete! Allah Hafiz.")
        break
    else:
        # Agar user ghalat option de to error message print kar rahe hain
        print("Invalid choice. Try again.")
```
